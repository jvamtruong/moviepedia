```python
import json

with open('movies_v2_data_response.jsonl', 'r', encoding='utf-8') as f:
  for line in f:
    doc = json.loads(line)
    index_details = {
      'index': {
        '_index': doc['_index'],
        '_id': doc['_id'],
      }
    }
    doc_body = {
      'title': doc['_source']['title'],
    }
    with open('./init/movies_v2_data_bulk.jsonl', 'a', encoding='utf-8') as f2:
      f2.write(json.dumps(index_details) + '\n')
      f2.write(json.dumps(doc_body) + '\n')
```
